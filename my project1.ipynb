```python
%%writefile image_annotator.py
import cv2
import numpy as np
import tkinter as tk
from tkinter import filedialog, simpledialog
import os
import sys

def has_display():

    if sys.platform.startswith("win"):
        return True

    return os.environ.get("DISPLAY") is not None

if not has_display():
    print("GUI not available in this environment.")
    print("Use this environment only to build the .exe with PyInstaller.")
    raise SystemExit


image = None
tool = "line"
start_point = None


def load_image():
    global image
    path = filedialog.askopenfilename(
        filetypes=[("Image Files", "*.jpg *.png *.jpeg")]
    )
    if path:
        image = cv2.imread(path)
        cv2.imshow("Image Annotator", image)

def save_image():
    if image is not None:
        path = filedialog.asksaveasfilename(
            defaultextension=".png",
            filetypes=[("PNG", "*.png"), ("JPG", "*.jpg")]
        )
        if path:
            cv2.imwrite(path, image)

def set_tool(t):
    global tool
    tool = t

def mouse_events(event, x, y, flags, param):
    global start_point, image

    if image is None:
        return

    if event == cv2.EVENT_LBUTTONDOWN:
        start_point = (x, y)

        if tool == "text":
            text = simpledialog.askstring("Add Text", "Enter label:")
            if text:
                cv2.putText(
                    image, text, (x, y),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.7, (0, 255, 0), 2
                )
                cv2.imshow("Image Annotator", image)

    elif event == cv2.EVENT_LBUTTONUP:
        end_point = (x, y)

        if tool == "rectangle":
            cv2.rectangle(image, start_point, end_point, (255, 0, 0), 2)

        elif tool == "line":
            cv2.line(image, start_point, end_point, (0, 0, 255), 2)

        elif tool == "circle":
            radius = int(np.linalg.norm(
                np.array(start_point) - np.array(end_point)
            ))
            cv2.circle(image, start_point, radius, (0, 255, 255), 2)

        elif tool == "measure":
            cv2.line(image, start_point, end_point, (0, 255, 0), 2)
            distance = int(np.linalg.norm(
                np.array(start_point) - np.array(end_point)
            ))
            mid_point = (
                (start_point[0] + end_point[0]) // 2,
                (start_point[1] + end_point[1]) // 2
            )
            cv2.putText(
                image, f"{distance}px", mid_point,
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6, (0, 255, 0), 2
            )

        cv2.imshow("Image Annotator", image)

cv2.namedWindow("Image Annotator")
cv2.setMouseCallback("Image Annotator", mouse_events)

root = tk.Tk()
root.title("Image Annotator & Measurement Tool")

tk.Button(root, text="Load Image", command=load_image).pack(fill="x")
tk.Button(root, text="Draw Rectangle", command=lambda: set_tool("rectangle")).pack(fill="x")
tk.Button(root, text="Draw Line", command=lambda: set_tool("line")).pack(fill="x")
tk.Button(root, text="Draw Circle", command=lambda: set_tool("circle")).pack(fill="x")
tk.Button(root, text="Add Text", command=lambda: set_tool("text")).pack(fill="x")
tk.Button(root, text="Measure Distance", command=lambda: set_tool("measure")).pack(fill="x")
tk.Button(root, text="Save Image", command=save_image).pack(fill="x")

root.mainloop()
cv2.destroyAllWindows()
```
